### Numpy doesn't truely generates random number it uses seed to generate data; if we fix the seed then everytime we will get exact same random numbers
### It is helpful when we need to share our notebook with others and we want them to get the exact same random numbers

```python
np.random.seed(7)
seeded_random_num7= np.random.randint(3,12,size=(3,3))
print(seeded_random_num7)
print('==============')
np.random.seed(99911)
seeded_random_num999= np.random.randint(3,12,size=(3,3))
print(seeded_random_num999)
```

## Dot Product

```python
np.random.seed(10)
m1 = np.random.randint(1,10,size=(2,3))
m2 = np.random.randint(1,10,size=(2,3))
print(m1,' \n === \n',m2)
```

```python
m1.dot(m2.T) # to multiply using transpose of m2
```
