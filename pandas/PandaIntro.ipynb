### creating data

```python
# creating column with Panda Series; 
# Note: number of value entered in series is less than number of rows still Pandas accepted the data and filled it with NAN
car_missing_data['Seats']=pd.Series([4,5,6,1])
```

```python
car_missing_data
```

```python
car_missing_data.fillna(4,inplace=True)
car_missing_data
```

```python
# will return error because while creating column with normal array length has to be same as number of rows
car_missing_data['KM per Liter']=[8,3,1] 
```

```python
car_missing_data['KM per Liter']=[8,3,1,4,5,6] 
car_missing_data
```

```python
car_missing_data['total fuel used']=car_missing_data['Odometer']/100 * car_missing_data['KM per Liter']
car_missing_data
```

```python
car_missing_data['number of wheels']=5
car_missing_data
```

```python
car_missing_data.drop('number of wheels',axis=1)
```

```python
# this is used to get random data and frac is used to get a portion of the entire data -> 1 being the max value 
# in this getting 50% of the data
shuffledData=car_missing_data.sample(frac=0.5) 
shuffledData
```

```python
shuffledData.reset_index(drop=True) #by default drop=FALSE because of that it will create a new column in the dataframe; to avoid it we used drop=true
```

```python
car_missing_data
```

```python
#apply helps to loop over data and apply value
car_missing_data['Odometer in miles'] = car_missing_data['Odometer'].apply(lambda x:x/1.6)
car_missing_data
```

```python
def categorize_seats(seats):
    print('SEATS====>',seats)
    if seats == 2:
        return 'SPORTS'
    elif seats == 4:
        return 'COUPE'
    elif seats == 5:
        return 'SEDAN'
    elif seats == 6:
        return 'MPV'
    elif seats == 7:
        return 'SUV'
    else:
        return 'VAN'

car_missing_data['Category']=car_missing_data['Seats'].apply(categorize_seats)
car_missing_data
```
