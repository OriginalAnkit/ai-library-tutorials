## Complete Example

```python
import matplotlib.pyplot as plt
x=[1,2,3,4,15]
y=[10,20,30,40,59]
fig,ax=plt.subplots(figsize=(20,10)) ## width , height
ax.plot(x,y)
ax.set(title="Sample Plot",xlabel='x-axis',ylabel='y-label')
```

```python
import numpy as np
```

```python
lineelements = np.linspace(1,10,100) # create 100 numbers with equal space in between
```

```python
fig,ax=plt.subplots(figsize=(5,5))
ax.plot(lineelements,lineelements**9); # by default creates a line graph
```

```python
# Scatter plots
fig,ax=plt.subplots(figsize=(5,5))
ax.scatter(lineelements,lineelements**9)
```

```python
fig,ax=plt.subplots(figsize=(5,5))
ax.scatter(lineelements,np.sin(lineelements))
```

```python
# Bar graph
butter_sales={"JAN":300,"FEB":800,"MAR":140,"APR":310,"MAY":100}
fig,ax=plt.subplots(figsize=(5,5))
ax.bar(butter_sales.keys(),butter_sales.values())
ax.set(title="Butter sales",xlabel="Month",ylabel="Sales in INR")
```

```python
butter_sales={"JAN":300,"FEB":800,"MAR":140,"APR":310,"MAY":100}
fig,ax=plt.subplots(figsize=(5,5))
ax.barh(butter_sales.keys(),butter_sales.values())
ax.set(title="Butter sales",xlabel="Month",ylabel="Sales in INR")
```

```python
# Histogram
x=np.random.randn(1000)
fig,ax=plt.subplots(figsize=(4,4))
ax.hist(x);
```

# SUBPLOTS

```python
fig,((ax1,ax2),(ax3,ax4))=plt.subplots(ncols=2,nrows=2,figsize=(10,5))
ax1.plot(lineelements,lineelements**2);
ax2.bar(lineelements,lineelements)
ax3.scatter(lineelements,lineelements**2)
ax4.barh(lineelements,lineelements**2)
```

```python
fig,((ax1,ax2),(ax3,ax4),ax5)=plt.subplots(ncols=2,nrows=3,figsize=(10,5))
ax1.plot(lineelements,lineelements**2);
ax2.bar(lineelements,lineelements)
ax3.scatter(lineelements,lineelements**2)
ax4.barh(lineelements,lineelements**2)
```
